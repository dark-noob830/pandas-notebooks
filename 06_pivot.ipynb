```python
import pandas as pd
import numpy as np
```

```python
ds = pd.read_csv('./data/sales_by_employee.csv',parse_dates=["Date"],date_format='%m/%d/%y')
ds
```

### Pivot_table

```python
ds.pivot_table(index='Date',aggfunc='sum',values='Expenses')
```

```python
ds.pivot_table(index='Date',aggfunc='sum',values=['Expenses','Revenue'])
```

```python
ds.pivot_table(index='Date',aggfunc='mean',values=['Expenses','Revenue'])
```

```python
ds.pivot_table(index='Date',aggfunc='sum',values=['Expenses','Revenue'],columns='Name')
```

```python
ds.pivot_table(index='Date',aggfunc='sum',values=['Expenses','Revenue'],columns='Name',margins=True)
```

```python
ds.pivot_table(index='Date',aggfunc='sum',values=['Expenses','Revenue'],columns='Name',margins=True,margins_name='Total')
```

```python
ds.pivot_table(index='Date',aggfunc='sum',values=['Expenses','Revenue'],columns='Name',margins=True,margins_name='Total',fill_value=0)
```

```python
ds.pivot_table(index='Date',aggfunc='sum',values='Revenue',columns='Name')
```

```python
ds.pivot_table(index='Date',aggfunc=['sum','count'],values='Revenue',columns='Name')
```

```python
ds.pivot_table(index='Date',aggfunc=['sum','mean'],values=['Expenses','Revenue'],columns='Name')
```

```python
ds.pivot_table(index='Date',aggfunc={'Expenses':'mean','Revenue':'count'},values=['Expenses','Revenue'],columns='Name')
```

```python
ds.pivot_table(index=['Name','Date'],aggfunc='sum',values='Revenue')
```

```python
ds.pivot_table(index=['Date','Name'],aggfunc='sum',values='Revenue')
```

```python
ds.pivot_table(index=['Date','Name'],aggfunc='sum',values='Revenue').index
```

### Stack/Unstack

```python
new = ds.pivot_table(index=['Name'],aggfunc='sum',values='Revenue',columns='Date')
new
```

```python
new.stack()
```

```python
new = ds.pivot_table(index=['Name','Customer'],aggfunc='sum',values='Revenue')
new
```

```python
new.unstack()
```
