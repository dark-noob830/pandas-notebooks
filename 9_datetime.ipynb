```python
dlv = pd.read_csv('data/deliveries.csv')
dlv
```

```python
dlv['order_date'] = pd.to_datetime(dlv['order_date'], format='%m/%d/%Y')
dlv['delivery_date'] = pd.to_datetime(dlv['delivery_date'], format='%m/%d/%Y')
dlv
```

```python
dlv['duration'] = dlv['delivery_date'] - dlv['order_date']
dlv
```

```python
dlv.sort_values('duration')
```

```python
dlv.sort_values('duration', ascending=False)
```

```python
dlv['duration'].mean()
```

```python
dlv[dlv['duration']>pd.Timedelta(days=365)]
```

```python
dlv[dlv['duration']>'365 days']
```
