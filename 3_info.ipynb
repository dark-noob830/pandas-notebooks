## 1) Optimize dtype and columns

```python
ep = pd.read_csv('data/employees.csv',parse_dates=['Start Date'],date_format="%m/%d/%y")
ep.info()
```

```python
ep['Mgmt'] = ep['Mgmt'].astype(bool) # Nan --- > True
ep.info()
```

```python
ep['Salary'] = ep['Salary'].fillna(0).astype('int32')
ep.info()
```

```python
ep.nunique()
```

```python
ep['Gender'] = ep['Gender'].astype('category')
ep.info()
```

```python
ep['Team'] = ep['Team'].astype('category')
ep.info()
```

## 2) Use Some Columns

```python
ep = pd.read_csv('data/employees.csv',parse_dates=['Start Date'],date_format="%m/%d/%y",usecols=['Start Date','Gender','Salary','Mgmt','Team'])
ep
```

```python
ep.info()
```
